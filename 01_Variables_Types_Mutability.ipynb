## What is a Variable?

A symbolic name that stores a value(s) in memory


## Why Do We Need Variables?

Without variables we can't:

*   Store data
*   Manipulate data
*   Analyze datasets
*   Build machine learning models

## Why This Matters in Data Analytics

* Variables are key to the data import phase

* Data cleaning relies on correct data types

* Machine learning workflows depend on reliable data handling

* Understanding mutability helps prevent difficult debugging problems

## Variable Types: Different kinds of data are stored using different types.

* integer (int): Whole numbers
* float (float): Number with decimals
* string (str): Text data
* boolean (bool): True or False values
* list (list): Ordered collection that can be modified (mutuble)
* tuple (tuple): Ordered collection that cannot be modified (immutable)
* dictionary (dict): Stores data as key-value pairs
* set (set): Stores unique values only

## Mutability

* Whether or not individual values of an object can be changed after it is created

* Mutable objects
  -can change individual values/characters
    *  list
    *  dictionary
    *  set

* Immutable objects
  -cannot change individual values/characters in a string
    *  integers
    *  strings
    *  tuples

## Sample Variable Types

```python
text = "I love learning business data analytics, especially with Dr. Marlin!!"
number = 10
decimal = 3.14
flag = True

print(text," : ", type(text))
print(number, " : ", type(number))
print(decimal, " :", type(decimal))
print(flag, " : ", type(flag))
```

### Common Variable Type Error
*  Data imported from CSV files is often stored as text when we expect numbers.

```python
age = "25"

try:
    print(age + 5)
except Exception as e:
    print("Error:", e)
```

Fix:
Convert the string to an integer first.

```python
age = "25"
print(int(age) + 5)
```

### Examples of Mutable Variables

```python
#Lists can be modified after creation.
numbers = [1,2,3]
print(numbers)
numbers[0] = 100
print(numbers)
```

```python
#Dictionary can be modified
s = "hello world"
print(s)
replacements = {'h': 'H', 'o': 'O'}
result = ''.join(replacements.get(c, c) for c in s)
print(result)  # "HellO wOrld"
```

### Examples of Immutable Variables
*  trying to change them will cause an error

```python
#trying to change a string from "Cold" to "Bold" that will result in an error
text = "Cold"
print(text)
text[0] = "B"
print(text)
```

```python
#Tuples are like lists, but cannot be modified after creation.
tup = (1,2,3)
print(tup)
tup[0] = 100
print(tup)
```

```python
#Strings cannot be changed after they are created
name = "Matt"

try:
    name[0] = "B"
except Exception as e:
    print("Error:", e)
```

# Summary

*  Variables are symbolic names that stores a value(s) in memory
*  Without variables we can't store, manipulate, or analyze datasets
*  Different variable types have different properties (mutability, arithmetic operations, etc)
