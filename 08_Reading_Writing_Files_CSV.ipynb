## 1. Introduction - What is a CSV File, and why is it important?

A Comma separated values (CSV) is a plain text file that stores tabular data. This data is stored in columns and rows, and commas seperate the values. CSV files typically look like this:

```python
#Pet,Species,Age,AdoptionFee_USD
#Fulffy,Cat,3,60
#Dofus,Dog,7,75
#Perry,Bird,1,35
```

In a CSV file, the first row containes the headers, and each row after that is a record. These are similar to Excel files, and can be opened in Excel, but CSV files are not the same as Excel files.

Why are these helpful? Lots of software and data is cateloged this way, and knowing how to work with it is expremtly essental to data analytics.

## 2. Reading a CSV File

First, we need a file to read. This cell creates pets.csv from our example. In an actual project, you might download a CSV instead.

```python
# Sample CSV
from pathlib import Path

csv_text = """Pet,Species,Age,AdoptionFee_USD
Fluffy,Cat,3,60
Dofus,Dog,7,75
Perry,Bird,1,35
"""
Path("pets.csv").write_text(csv_text, encoding="utf-8")
print("Created pets.csv")
```

```python
import pandas as pd

# Read the CSV file into a table called a DataFrame
pets = pd.read_csv("pets.csv")
display(pets)
```

`pd.read_csv()` opens the file, and `display()` shows the data as a table. Pandas is a Python library used to work with tables and datasets.

# A few other ways to read a CSV
We have already read the whole pets.csv file. Sometimes we only need part of a file. Here are a few other options using the same example.

Read just the first two rows: This is useful if we want a quick look at a large file.

```python
# nrows=2 tells pandas to read only the first two data rows
first_two = pd.read_csv("pets.csv", nrows=2)
display(first_two)
```

Read only certain columns: This is helpful when a CSV has a lot of columns and we do not need all of them.

```python
# usecols lets us choose which columns to bring in
just_fees = pd.read_csv("pets.csv", usecols=["Pet", "AdoptionFee_USD"])
display(just_fees)
```

# 3. Writing a CSV File

Reading a CSV brings data into Python. Writing a CSV saves data from Python into a file. With pandas, we use `.to_csv()` to do that.

Here are a few ways we can write CSV files using the same pet example.

Create a new CSV
First, let's make a small table for a new pet, then save it as its own CSV file.

```python
# Create a table with one pet
new_pet = pd.DataFrame({
    "Pet": ["Buddy"],
    "Species": ["Dog"],
    "Age": [2],
    "AdoptionFee_USD": [65]
})

new_pet.to_csv("new_pet.csv", index=False)
print("Saved new_pet.csv")
display(pd.read_csv("new_pet.csv"))
```

Notice the `index=False` part. Without it, pandas would also save the row numbers as a separate column. We usually don't need those in the CSV.

**Save changes to an existing table**

We can also change some data and save the updated version as a new file. Here, we'll change Perry's adoption fee from $35 to $40. The original pets.csv will stay the same.

```python
# Make a copy so our original table stays the same
pets_updated = pets.copy()
pets_updated.loc[pets_updated["Pet"] == "Perry", "AdoptionFee_USD"] = 40

pets_updated.to_csv("pets_updated.csv", index=False)
print("Saved pets_updated.csv")
display(pd.read_csv("pets_updated.csv"))
```

**What happens if we use the same file name?**

If a file already exists, .to_csv() will replace it when we save using the same name. Let's try that with the small new_pet.csv file we just made, so we don't overwrite our original pet data.

```python
# Change Buddy's adoption fee and save using the same file name
new_pet.loc[0, "AdoptionFee_USD"] = 70
new_pet.to_csv("new_pet.csv", index=False)

# Read the file again to see the new value
display(pd.read_csv("new_pet.csv"))
```

The file is still called `new_pet.csv`, but Buddy's fee is now 70 instead of 65. If we want to keep both versions, we need to save the updated table under a **different file name**.

Later, we'll use the same writing method to save just the rows we want from the pet and Census data.

# 4. File Paths and Common Mistakes
A file path tells Python where a file is saved. In this section we show 3 common mistakes people make when reading and writing CSV files. For each one we show the error, explain why it happens, and then fix it.

# Checking where Colab is looking
Google Colab normally starts in the `/content/ `folder. We can check our current folder and see the files that are saved there. If Python cannot find a file, this is one of the first things to check.

```python
import os

print("Current folder:", os.getcwd())
print("Files in this folder:", os.listdir())
```

## Mistake 1: FileNotFoundError

**Quick question:** Our file is called `pets.csv`. What happens if we accidentally type `pet.csv` instead? Let's try it below.

We use `try` / `except` so the notebook keeps running after the error. Python tries the code under `try`. If it gets a FileNotFoundError, it runs the `except` part and prints our message instead of stopping.

```python
# Wrong file name: pet.csv instead of pets.csv
try:
    pd.read_csv("pet.csv")
except FileNotFoundError:
    print("FileNotFoundError: Python could not find 'pet.csv'.")

# Check the real file names
print("Files in this folder:", os.listdir())
```

The same error can happen when the file name is correct but the folder is wrong. Below, we try the wrong folder first and then use the correct path.

```python
# There is no "data" folder, so Python can't find the file
try:
    pd.read_csv("data/pets.csv")
except FileNotFoundError:
    print("FileNotFoundError: 'pets.csv' is not in the 'data' folder.")

# Fix: give the full path where the file really is
# /content is Colab's main folder, and pets.csv is inside it
pets = pd.read_csv("/content/pets.csv")
display(pets)
```

## Mistake 2: the "Unnamed: 0" column

We have already seen this in class. In Lesson 14, when we loaded `new_donors.csv`, the first column was `Unnamed: 0` and it only had the numbers 0 to 9.

Here we save the pets table but forget `index=False`.

```python
# Save without index=False
pets.to_csv("pets_v2.csv")

# Read it back
pets_v2 = pd.read_csv("pets_v2.csv")
print("Columns:", pets_v2.shape[1])
display(pets_v2)
```

### Why it happens
Every pandas table has row numbers on the left (0, 1, 2...), called the index. They are like the row numbers in Excel, not real data. If we forget `index=False`, pandas saves those row numbers into the file as a real column. That column has no name, so when we read the file back pandas calls it `Unnamed: 0`.

Our table went from 4 columns to 5. If this extra column goes into a model, the model tries to learn from row numbers, which mean nothing.

### How to fix it
Save with `index=False`.

```python
# Save again, this time with index=False
pets.to_csv("pets_v2.csv", index=False)

pets_v2 = pd.read_csv("pets_v2.csv")
print("Columns:", pets_v2.shape[1])
display(pets_v2)
```

## Mistake 3: zip codes lose their leading 0

Our pets table has no zip codes, so here is a small table of 4 animal shelters. 2 of the 4 zip codes start with 0 (Boston and Portland, Maine).

```python
shelters = pd.DataFrame({
    "Shelter": ["Boston Shelter", "Portland Shelter", "Richmond Shelter", "New York Shelter"],
    "Zip": ["02134", "04101", "23220", "10001"]
})

# Save it and read it back
shelters.to_csv("shelters.csv", index=False)
shelters_loaded = pd.read_csv("shelters.csv")

display(shelters_loaded)
print(shelters_loaded.dtypes)
```

### Why it happens
`02134` became `2134` and `04101` became `4101`. `Zip` now says `int64`, which means pandas turned it into a number.

A CSV is only text, so it doesn't save the type of each column. When pandas reads the file it guesses the type. `02134` is all digits, so pandas makes it the number 2134, and numbers don't keep a 0 at the start.

There is no error message, so it is easy to miss. 2 of our 4 shelters (50%) lost their leading 0. If we search or join tables by zip code, Boston and Portland won't match anymore.

```python
# Tell pandas that Zip is text, not a number
shelters_fixed = pd.read_csv("shelters.csv", dtype={"Zip": str})

display(shelters_fixed)
print(shelters_fixed.dtypes)
```

The 0s are back: `02134` and `04101`. Do the same for any ID that is made of digits but is not really a number, like zip codes, phone numbers and account numbers.

# 5. Using this Data

Once we have read the CSV, we can work with the data. For example, let’s find the pets with adoption fees over $50, then save those results to a new CSV file.

```python
# Keep only pets with adoption fees over $50
pets_over_50 = pets[pets["AdoptionFee_USD"] > 50]
display(pets_over_50)

# Save the results as a new CSV
pets_over_50.to_csv("pets_over_50.csv", index=False)
print("Saved pets_over_50.csv")
```

# 6. Why it Matters
In data analytics, we often start with a CSV from a report, a survey, or a public source like the Census Bureau. We need to read it correctly before analyzing it, then write a new file to save or share our results.

| Mistake | What goes wrong | Do you see an error? |
|---|---|---|
| FileNotFoundError | Nothing runs | Yes, right away |
| Unnamed: 0 column | An extra column gets added (4 columns became 5). A model can end up learning from row numbers | No |
| Leading zeros lost | 2 of our 4 shelters (50%) lost their 0, so searching or joining by zip misses them | No |

Mistake 1 stops your code, so you notice it and fix it. Mistakes 2 and 3 don't stop anything. The code finishes and the numbers are wrong, and those wrong numbers can end up in a report.

## 3 habits to remember
1. Check where you are and what files are there: `os.getcwd()` and `os.listdir()`
2. Save with `index=False`
3. After every `read_csv`, look at the table and `.dtypes` before you do anything else
