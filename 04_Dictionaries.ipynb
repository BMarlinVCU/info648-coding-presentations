Definition of Dictionary

A dictionary is a Python data structure that stores information as key-value pairs. Unlike a list, where you retrieve information using its position, a dictionary lets you retrieve information using a meaningful label.

```python
#Example

student = {
    "name": "Ada",
    "program": "Data Science",
    "gpa": 3.9
}
```

Think of the dictionary as an entire student record:

| Key         | Value            |
| ----------- | ---------------- |
| `"name"`    | `"Ada"`          |
| `"program"` | `"Data Science"` |
| `"gpa"`     | `3.9`            |


The { } curly brackets indicate a dictionary.


```python
#Business example
#Imagine information about a shipment:

shipment = {
    "shipment_id": "SH1001",
    "carrier": "ABC Freight",
    "cost": 2500,
    "status": "In Transit"
}
```

```python
#Instead of remembering where the shipping cost is stored, you can simply ask:
shipment["cost"]
```

Definition of a Key

A key identifies a piece of information.

```python
#Example
shipment = {
    "carrier": "ABC Freight",
    "cost": 2500
}

#"carrier" and "cost" are keys.
```

Think of a key as an Excel column heading:

| Excel Column | Python Key  |
| ------------ | ----------- |
| Carrier      | `"carrier"` |
| Cost         | `"cost"`    |
| Status       | `"status"`  |


```python
#When you write:
shipment["carrier"]

#you are essentially telling Python:

#Go to the shipment dictionary, find the key called "carrier", and give me its value.
```

Keys must be hashable, meaning their hash value cannot change during their lifetime. Strings, numbers, and appropriate tuples can therefore be keys, while mutable objects such as lists and dictionaries generally cannot. Keys are also unique; assigning another value to an existing key replaces the old value.

Definition of Values

A value is the information associated with a key.

Example #1

"cost": 2500

Example # 2

"status": "In Transit"

Unlike keys, values can contain essentially any Python object, including numbers, strings, lists, functions, or even other dictionaries.

```python
shipment = {
    "shipment_id": "SH1001",
    "carrier": "ABC Freight",
    "cost": 2500,
    "stops": ["Richmond", "Charlotte", "Atlanta"]
}

#Here "stops" is the key, but its value is an entire list.
```

Dictionaries Can Contain Other Dictionaries

This is where dictionaries become especially useful in business applications.

Suppose an employee record contains compensation information:

```python
employee = {
    "name": "Maria",
    "department": "Finance",
    "compensation": {
        "salary": 85000,
        "bonus": 5000
    }
}

#The value associated with "compensation" is another dictionary.
```

```python
#To retrieve salary:

employee["compensation"]["salary"]
```

Think about the command one step at a time:

    employee

    ↓
    compensation

    ↓
    salary

    ↓
    85000

This is called a nested dictionary.

## Connecting APIs, JSON, and Dictionaries

An API allows two systemns to exchange information. For example, an HR system's API could return employee hire-date information in JSON format.
Python can convert the JSON dictionaries for analysis.

```python
employee_data = {
    "employees": [
        {
            "employee_id": 101,
            "department": "Finance",
            "hire_date": "2022-06-13"
        },
        {
            "employee_id": 102,
            "department": "Sales",
            "hire_date": "2024-01-08"
        },
        {
            "employee_id": 103,
            "department": "Finance",
            "hire_date": "2024-09-15"
        }
    ]
}

employee_data
```

## Preparing the Informationh for Analysis

The list of employee dictionaries can be converted into a Pandas DataFrame. Each employee becomes a row and the dictionary keys become columns.

```python
import pandas as pd

employees_df = pd.DataFrame(employee_data["employees"])
employees_df
```

## Common Mistake: Using the Wrong Key

Dictionary keys must match exactly. Using `"Hire_Date"` instead of `"hire_date"` would cause a KeyError because keys are case-sensitive.

```python
employee_data["employees"][0].get("Hire_Date","Key not found")
```

```python
employee_data["employees"][0]["hire_date"]
```

## Key Takeaway

The API retrieves the information, JSON organizes it for transmission, and Python dictionaries allow analysts to access and prepare the information for analysis.
