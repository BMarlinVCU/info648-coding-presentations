#Running out of memory

**What Is a Memory Error?**

Python uses RAM (Random Access Memory) to temporarily store variables, datasets, lists, arrays, and other objects while a program runs. A "MemoryError" occurs when Python attempts to use more RAM (Random Access Memory) than is available.

Common Culprits:
*   Loading very large datasets into Pandas.
*   Creating multiple copies of large DataFrames.
*   Storing millions of values in lists or arrays.
*   Processing more information than the computer can handle at once.

```python
#Increasing the size of a list increases memory usage:

import sys

small_list = [0] * 1000
large_list = [0] * 1000000

print("Small list:", sys.getsizeof(small_list), "bytes")
print("Large list:", sys.getsizeof(large_list), "bytes")
```

**Option 1. Use More Efficient Data Types**

Pandas may use more memory than necessary when storing numerical data. For example, int64 and float64 use more memory than int32 and float32.

```python
import pandas as pd

df = pd.DataFrame({
    "ID": range(1000000),
    "Income": [50000.0] * 1000000
})

# Check original data types and memory
print("Original data types:")
print(df.dtypes)

# Reduce data types
df["ID"] = df["ID"].astype("int32")
df["Income"] = df["Income"].astype("float32")

# Check updated data types and memory
print("\nUpdated data types:")
print(df.dtypes)
```

Smaller data types require less memory per observation.

Converting int64 to int32 reduces the storage required for each integer by half.

Converting float64 to float32 also reduces storage by half, although it provides less numerical precision.

**Option 2. Process Large Datasets in Chunks**

When working with large datasets, loading everything into memory at once can cause problems. Instead, process smaller portions of the dataset at a time.

```python
#create a sample dataset:

import pandas as pd

df = pd.DataFrame({
    "ID": range(100000),
    "Income": range(100000),
    "Age": range(100000)
})

df.to_csv("sample_data.csv", index=False)

#compare loading the entire dataset with processing it in chunks:

# Load entire dataset
df = pd.read_csv("sample_data.csv")

print("Entire dataset:", df.shape)

# Process dataset in smaller portions
for chunk in pd.read_csv("sample_data.csv", chunksize=10000):
    print("Chunk:", chunk.shape)
```

The first approach loads all 100,000 observations into memory simultaneously.

The second approach processes 10,000 observations at a time.

Each chunk can be processed and discarded before moving to the next one.


**Option 3. Avoid Unnecessary Copies**

Creating copies of large DataFrames can significantly increase memory consumption.

Instead of:

df_copy = df.copy()

If you do not need an independent copy, simply reference the original:

df_copy = df

Both variables now reference the same DataFrame, rather than storing two separate copies.

**Key Takeaway**

First, investigate how your code is using memory. Often, you can resolve the problem by:

*   Loading less data
*   Avoiding unnecessary copies
*   Processing data in smaller chunks

The goal is to make your code more memory-efficient before increasing your computing resources.
