# **Lists vs. Tuples**

## What are Lists and Tuples?

Both lists and tuples are ordered collections in Python.

**Lists**
- Use square brackets: `[]`
- Ordered and indexed
- Mutable, meaning they can be changed after creation
- Useful when values may need to be added, removed, or updated

**Tuples**
- Use parentheses: `()`
- Ordered and indexed
- Immutable, meaning they cannot be changed after creation
- Useful when values should stay fixed

```python
test_list = [1, 2, 3, 4, 5]
test_tuple = (1, 2, 3, 3, 4, 5)

print("List:", test_list)
print("Tuple:", test_tuple)
```

## Mutability vs. Immutability

A list can be changed after it is created.

A tuple cannot be changed after it is created.

For example, `.append()` works on a list, but tuples do not have an `.append()` method.

```python
test_list.append(6)

print("Updated list:", test_list)
print("Tuple stays the same:", test_tuple)
```

## Indexing

Python starts counting positions at 0.

```python
colors_list = ["red", "blue", "green"]
colors_tuple = ("red", "blue", "green")

print("List index 0:", colors_list[0])
print("Tuple index 0:", colors_tuple[0])

print("List index 2:", colors_list[2])
print("Tuple index 2:", colors_tuple[2])
```

### Count

The `.count()` method tells us how many times a specific value appears in a tuple.

```python
print("The number 3 appears", test_list.count(3), "times in the list.")
print("The number 3 appears", test_tuple.count(3), "times in the tuple.")
```

## Quick Comparison

| Feature | List | Tuple |
|---|---|---|
| Syntax | `[]` | `()` |
| Ordered | Yes | Yes |
| Indexed | Yes | Yes |
| Mutable | Yes | No |
| Can use `.append()` | Yes | No |
| Best when | Data may change | Data should stay fixed |

## Why This Matters in Data Analytics

**Lists**
- Useful when data may change during cleaning or analysis
- Example: collecting monthly sales values, adding observations, or removing bad data

**Tuples**
- Useful when values should remain fixed
- Example: latitude/longitude coordinates, database records, or fixed configuration values

Choosing the wrong structure can cause errors or make data easier to change accidentally.

## Common Mistake 1: Forgetting the Comma in a One-Item Tuple

The mistake: A student tries to create a tuple with one item, but Python reads the value as a string or number instead.

Why it occurs: Parentheses alone do not create a one-item tuple. Python needs a trailing comma to recognize it as a tuple.

```python
not_a_tuple = ("customer_id")
one_item_tuple = ("customer_id",)

print(type(not_a_tuple))
print(type(one_item_tuple))
```

## Common Mistake 2: Trying to Change a Tuple

The mistake: A tuple is treated like a list and an item is changed using its index.

Why it occurs: Lists and tuples can both be indexed, so they look similar when accessing values. However, tuples are immutable, so their values cannot be changed after creation.

This tuple contains the user's ID, date, and status:

`record = ("User_102", "2026-03-10", "Active")`

This would cause an error:

`record[2] = "Inactive"`

The value at index `2` is the third item, `"Active"`. Because `record` is a tuple, Python will not allow us to replace it.

```python
record = ("User_102", "2026-03-10", "Active")

record_list = list(record)
record_list[2] = "Inactive"
record = tuple(record_list)

print(record)
```

## Common Mistake 3: Using a List as a Dictionary Key

**The mistake:** A student tries to use a list as the key in a dictionary.

**Why it occurs:** Lists can change, so Python does not allow them to be used as dictionary keys. Tuples cannot be changed, so they can be used as keys.

```python
store_location = (40.7128, -74.0060, "Main Street")

location_sales = {store_location: 25000}

print(location_sales)
```

This will cause an error because lists are mutable:

`store_location = [40.7128, -74.0060]`

`location_sales = {store_location: 25000}`

## Memory: Why Mutability Matters

Lists are mutable, so Python can change the contents of the existing list.

Tuples are immutable, so Python cannot change the original tuple. If we "change" a tuple, Python creates a new tuple instead.

```python
print("List before change:", test_list)
print("List ID before:", id(test_list))

test_list.append(7)

print("List after change:", test_list)
print("List ID after:", id(test_list))
```

The list changed, but its `id()` stayed the same.

This shows that Python modified the existing list object.

```python
print("Tuple before change:", test_tuple)
print("Tuple ID before:", id(test_tuple))

test_tuple = test_tuple + (7,)

print("Tuple after change:", test_tuple)
print("Tuple ID after:", id(test_tuple))
```

The tuple now contains another value, but Python did not modify the original tuple.

Because tuples are immutable, `test_tuple + (7,)` creates a new tuple and assigns the name `test_tuple` to that new object. The `id()` changes because it is a different object.
