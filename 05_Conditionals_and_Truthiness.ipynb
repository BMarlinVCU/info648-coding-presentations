# Part 1: Python Conditionals

## What Are Conditionals?

Conditionals allow Python to make decisions based on whether a condition is `True` or `False`.

Python uses three main keywords:
- `if`: Checks the first condition.
- `elif`: Checks another condition.
- `else`: Runs when none of the previous conditions are True.

Conditionals are commonly used in data analysis to classify data and apply business rules.

```python
if condition_1:
    pass  # Runs if condition_1 is True
elif condition_2:
    pass  # Runs if condition_1 is False
          # and condition_2 is True
else:
    pass  # Runs if all previous conditions are False
```

## Basic Example: Student Grades

````python
score = 85

if score >= 90:
    print("Grade A")
elif score >= 80:
    print("Grade B")
elif score >= 70:
    print("Grade C")
else:
    print("Grade F")
````

## Real-World Example: Customer Segmentation

Companies can use conditional statements to classify customers based on spending.

Business Rules:
- VIP: Spending >= 1,000
- Regular: Spending >= 500
- Basic: Spending < 500

````python
# Customer Segmentation

spending = 1200

if spending >= 1000:
    customer_level = "VIP"
elif spending >= 500:
    customer_level = "Regular"
else:
    customer_level = "Basic"

print("Customer Level:", customer_level)
````

## Common Mistake: Incorrect Condition Order

Python evaluates conditions from top to bottom.

If a broader condition appears before a more specific condition, the result may be incorrect.

````python
# Incorrect Example

spending = 1200

if spending >= 500:
    print("Regular")
elif spending >= 1000:
    print("VIP")
else:
    print("Basic")
````

````python
# Correct Example

spending = 1200

if spending >= 1000:
    print("VIP")
elif spending >= 500:
    print("Regular")
else:
    print("Basic")
````

# Part 2: Truthiness

## What Is Truthiness?

Every Python value is either **truthy** or **falsy** when used in an `if`.

**Falsy:** `False`, `None`, `0`, `0.0`, `""`, `[]`, `{}`, `()`, `set()`

**Everything else is truthy** — including `"0"`, `[0]`, `(0,)`, and negative numbers.

````python
# Quick check with bool()

print(bool(0), bool(1))         # False True
print(bool(""), bool("0"))      # False True  <- "0" is a non-empty string!
print(bool([]), bool([0]))      # False True  <- non-empty list, even with a falsy item
print(bool(None))               # False
print(bool((0)), bool((0,)))    # False True  <- (0) is just 0; (0,) is a real tuple
````

## Real-World Example: Filtering Customer Feedback

Say we're processing customer feedback comments, and we only want to act on the ones where a comment was actually written.

````python
# Filtering out empty feedback using truthiness

feedback_list = ["Great service!", "", "Could be faster", None, "Loved it"]

for comment in feedback_list:
    if comment:
        print("Processing comment:", comment)
    else:
        print("Skipped empty comment")
````

## Common Mistake: `0` Is Not the Same as `None`

Using `if not value:` treats `0` and `None` the same way — both are falsy. But `0` is a real value (e.g., "0% discount"), while `None` means nothing was ever set. Use `is None` when you need to tell them apart.

````python
# Incorrect Example

percent_off = 0

if not percent_off:
    print("No discount applied")
else:
    print("Discount applied:", percent_off, "%")

# Bug: prints "No discount applied" even though percent_off really is 0%
````

````python
# Correct Example

percent_off = 0

if percent_off is None:
    print("No discount applied")
else:
    print("Discount applied:", percent_off, "%")

# Now correctly prints "Discount applied: 0 %"
````

# Summary

- **Conditionals** (`if` / `elif` / `else`) let code branch based on business rules — order the conditions from most specific to least specific.
- **Truthiness** lets any value act like `True`/`False` — but watch out for `0`, `""`, and `[]` being falsy even when they're meaningful data.
- Both are foundational for data cleaning, filtering, and building analytics pipelines with Pandas.
