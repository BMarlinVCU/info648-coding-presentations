## 1. What Does "Writing Python for Humans" Mean?

Python's philosophy is built on readability. Python code isn't just instructions for a computer—it is a document that human teammates must read, debug, and maintain.

### The Standard: PEP 8
**PEP 8** is Python's official style guide. It sets rules for formatting so code looks uniform across teams:
- **Variable & Function Names:** Use `snake_case` (all lowercase with underscores).
- **Indentation:** Use **4 spaces** per indent level (no mixing tabs and spaces).
- **Constants:** Use `ALL_CAPS` for fixed settings (e.g., `TAX_RATE = 0.07`).
- **Whitespace:** Use spaces around operators (`total = price + tax`, not `total=price+tax`).

## Part 1: Meaningful Names

A variable stores a value. A function is a named, reusable calculation: it receives inputs and returns a result.

**Before running:** What do `d`, `p`, and `t` mean in the first example?

The first version works, but its names give the reader little help. The second version makes the same calculation with clearer names.

```python
#BAD (Cryptic for humans)
d=7
def f(p,t):
    return p*(1+t)

# GOOD (Self-explanatory for humans)
days_until_deadline = 7

def calculate_total_price(price, tax_rate):
    """Calculates final price including sales tax."""
    return price * (1 + tax_rate)

print(f"Days left: {days_until_deadline}")
print(f"Total: ${calculate_total_price(100, 0.07):.2f}")
```

Both versions calculate **107** from a price of **100** and a tax rate of **0.07**. The clearer version also labels the output as money.

`def` defines the function, `price` and `tax_rate` name its inputs, and `return` sends the calculated value back. The `:.2f` in the printed message displays two decimal places.

The result did not change when we improved the names. Names should explain the meaning; they do not have to be long. Familiar names such as `df` can be useful when their meaning is clear from the surrounding code.

## Part 2: Visual Style

Python uses indentation to group statements. PEP 8 recommends **four spaces** per indentation level; four is a style convention, while consistent block indentation is required by Python.

- **Line length:** PEP 8 generally recommends a maximum of 79 characters for code. Short lines are also easier to show on a classroom screen.
- **Spaces:** Use spaces around `=` and operators such as `+` and `>=`.
- **Blank lines:** Separate related steps so the reader can see the structure.

In this example, `if` checks whether the age is at least 18. Both versions work and print the same message. The spacing in the second makes the comparison easier to scan.

```python
# BAD (Cramped and messy)
user_age=20
if user_age>=18:
    print('Access granted')

# GOOD (Clean spacing, easy on the eyes)
user_age = 20

if user_age >= 18:
    print("Access granted")
```

## Part 3: Structuring Your Script

Organize a short script so a reader can follow it:
1. **Imports:** Make the tools you need available.
2. **Helper functions:** Define reusable tasks.
3. **Main execution code:** Supply inputs, call the functions, and show results.

The example below follows this order. `random.choice` selects one greeting from a list, so the greeting may differ each time you run it.

In an analysis notebook, use headings and short cells to guide readers through **data → preparation → calculation → results**. Run the cells from top to bottom so the required inputs are defined before they are used.

```python
# 1. IMPORTS
import random

# 2. HELPER FUNCTIONS
def get_random_greeting(name):
    greetings = ["Hello", "Welcome", "Hey there"]
    chosen_greeting = random.choice(greetings)
    return f"{chosen_greeting}, {name}!"

# 3. MAIN SCRIPT LOGIC
student_name = "Alex"
message = get_random_greeting(student_name)
print(message)
```

## Part 4: Useful Comments and Docstrings

- **Comments (`#`):** Explain a decision or business rule when the code alone does not make it clear. Avoid merely repeating an obvious step.
- **Docstrings:** Put a short description as the first statement inside a function. Explain its purpose and any input meaning the reader needs.
- **Keep them current:** When a rule changes, update its explanation too.

In the example below, the extra seat is for the instructor. The discount function returns the **final price**, so its name says `calculate_discounted_price`. `is_member` is either `True` or `False`.

```python
count = 0
student_count = 10

# BAD COMMENT: this repeats the operation without explaining its purpose.
# Add 1 to count.
count = count + 1

# GOOD COMMENT: this explains the reason for the extra place.
# Reserve one additional seat for the instructor.
total_capacity = student_count + 1

```

```python
def calculate_discounted_price(price, is_member):
    """Return the final price after applying the membership rule.

    Members receive a 20% discount; nonmembers pay the full price.
    """
    if is_member:
        return price * 0.80
    return price


print(f"Total capacity: {total_capacity}")
print(f"Member price: ${calculate_discounted_price(100, True):.2f}")
print(f"Nonmember price: ${calculate_discounted_price(100, False):.2f}")
```

## Part 5: A Common Mistake in an Analytics Calculation

**Scenario:** An analyst reads a price of **$100** and a tax percentage of **7** from a sales data table. Here, we enter those two values directly so the example needs no external file.

The function `calculate_total_price` expects a decimal tax rate. The analyst passes the percentage directly.

**Predict before running:** Should a 7% tax turn a $100 price into $107 or $800?

```python
# MISTAKE: passing a percentage where the function expects a decimal rate.
price = 100
tax_percent = 7

wrong_total = calculate_total_price(price, tax_percent)
print(f"Incorrect total: ${wrong_total:.2f}")
```

### Why the Mistake Occurs

The function calculates `100 * (1 + 7)`, which is **800**. It treats `7` as a decimal rate of **700%**, not **7%**.

The program runs without an error, but a sales report using this result would overstate the tax-inclusive amount. A short name such as `t` makes the unit easy to miss.

### How to Fix It

Divide the percentage by 100: **7 / 100 = 0.07**. Keep the two meanings visible with the names `tax_percent` and `tax_rate`.

```python
tax_rate = tax_percent / 100
correct_total = calculate_total_price(price, tax_rate)

print(f"Tax percentage: {tax_percent}%")
print(f"Decimal tax rate: {tax_rate}")
print(f"Correct total: ${correct_total:.2f}")
```

The corrected result is **$107.00**. The conversion fixes the calculation. Clear names and the function's docstring help the reader notice which form is needed.

This differs from our earlier style changes: better spacing and names kept the result the same; here, we corrected the value passed into the calculation.

We use two clear steps: first convert the percentage, then calculate the total. Keeping the steps separate makes the analysis easier for another person to follow and check.

## Summary Checklist for Human-Friendly Code
1. **Name items for humans:** `customer_age` beats `a` or `temp`.
2. **Follow PEP 8:** Use `snake_case` for functions/variables and **4 spaces** for indents.
3. **Structure logically:** Group imports at top, helper functions in middle, execution at bottom.
4. **Write code to be read:** Assume someone else will read your notebook tomorrow.
