## Part 1. Introduction to Strings

A string is a data type in Python used to store text. Strings can contain letters, numbers, spaces, and special characters.

Strings are created by placing text inside single (' ') or double (" ") quotation marks.

Strings are commonly used to store information such as names, addresses, messages, and identification numbers.


```
Program = 'Master of Decision Analytics'
Class = "INFO 648"
print(Program)
print(Class)
```

```
age = 25
age_text = "25"

print(type(age))
print(type(age_text))
```

##Indexing and Slicing


Indexing: Allows us to access an individual character in a string using its position.



Slicing: Allows us to extract a portion of a string by specifying a range of positions.



Python uses zero-based indexing, meaning the first character starts at position 0 instead of 1.

```

Program = "MDA"

print(Program[0])
print(Program[1])
print(Program[2])
```

```

Class = "INFO 648"

print(Class[0:4])
print(Class[5:8])
```

##String Concatenation

String concatenation is the process of joining two or more strings together.

In Python, we can use the + operator to combine strings.

```

Program = "MDA"
Class = "INFO 648"

print(Program + Class)
print(Program + " " + Class)
```

##Basic String Operations

Python provides several built-in functions and methods that allow us to work with strings.

Some common operations include:
- len() — Counts the number of characters in a string, including spaces.
- .upper() — Converts all letters in a string to uppercase.
- .lower() — Converts all letters in a string to lowercase.

```

Program = "MDA"
Class = "info 648"

print(len(Class))
print(Program.lower())
print(Class.upper())
```

## Splitting and Joining Strings

.split() — Separates a string into smaller pieces and stores them in a list.



.join() — Combines items from a list into one string using a separator.

By default, .split() separates text wherever there's whitespace, such as a space.

```

Class = "INFO 648"

# Splitting a string
print(Class.split())

# Joining strings
Class_list = ["INFO", "648"]
print(" ".join(Class_list))
```

.split() produces a list, while .join() produces a string.

#Searching Strings
##.find() and .index()

Python provides two methods for locating characters or text within a string:
- .find() — Searches for a character or substring and returns its starting index. If it isn't found, Python returns -1.
- .index() — Also searches for a character or substring and returns its starting index. However, if it isn't found, Python produces an error.
Both methods use zero-based indexing, meaning counting starts at 0.

```
Class = "INFO 648"

print(Class.find("6"))
print(Class.index("6"))
print(Class.find("9"))
```

##String Validation

Python provides several methods to check whether a string contains certain types of characters.
- .isalpha() — Checks if all characters are letters.
- .isdigit() — Checks if all characters are digits.
- .isalnum() — Checks if all characters are letters or numbers, without spaces or special characters.
These methods return a Boolean value: True or False.

```

Program = "MDA"
Class = "INFO 648"
Course_Number = "648"

print(Program.isalpha())
print(Class.isalnum())
print(Course_Number.isdigit())
```

String validation doesn't change the original string. It simply checks whether the string meets a particular condition.

#3. String Basics
## F-Strings

An f-string (formatted string literal) is a simple way to combine text and variables in Python.
- Place an f before the opening quotation mark.
- Use curly brackets {} to insert variables.
- Python automatically replaces the variables with their values.

```

Program = "MDA"
Class = "INFO 648"

print(f"I am enrolled in the {Program} program.")
print(f"One of my classes is {Class}.")
```

##Part 2: String Formatting: % Formatting

Percent formatting uses the % operator and special placeholders to insert values into a string.
- %s — Inserts a string (text).
- %d — Inserts an integer (whole number).
- %f — Inserts a floating-point number (decimal).
Although f-strings and .format() are more commonly used in modern Python, percent formatting can still be found in older code.

```

Program = "MDA"
Class = "INFO 648"
Credits = 3

print("I am enrolled in the %s program." % Program)
print("I am taking %s for %d credits." % (Class, Credits))
```

```

Program = "MDA"

# F-string
print(f"My program is {Program}.")

# .format()
print("My program is {}.".format(Program))

# Percent formatting
print("My program is %s." % Program)
```

##Number formatting

```

Salary = 57500.567
Savings_Rate = 0.25

print(f"My salary is ${Salary:.2f}")
print(f"My salary is ${Salary:,.2f}")
print(f"My savings rate is {Savings_Rate:.1%}")
```

##Date and Time Formatting

Python provides a module called datetime that allows us to work with dates and times.

We can use the .strftime() method to format dates into readable strings.

Some common formatting codes include:
- %Y — Four-digit year (2026)
- %m — Month as a number (11)
- %d — Day of the month (17)
- %B — Full month name (November)
- %A — Full weekday name (Tuesday)

```

from datetime import datetime

Date = datetime(2026, 11, 17)

print(Date.strftime("%m/%d/%Y"))
print(Date.strftime("%B %d, %Y"))
print(Date.strftime("%A, %B %d, %Y"))
```

##Escape Characters

Escape characters allow us to include special formatting within strings.

In Python, escape sequences begin with a backslash (\).

Common escape sequences include:
- \n — Creates a new line.
- \t — Adds a tab space.
- \" — Inserts double quotation marks within a string.
- \\ — Inserts a backslash.

```

Program = "MDA"
Class = "INFO 648"

print("Program:\nMDA")
print("Class:\tINFO 648")
print("I am enrolled in the \"MDA\" program.")
```

## ⚠️ Common string mistakes

When working with strings in Python, some common mistakes include:

- Combining strings and integers: Python cannot directly concatenate a string and an integer using +.
- Incorrect indexing: Python starts counting at 0, not 1.

```

# Correct use of quotation marks
Program = "MDA"

# Converting an integer to a string
Credits = 3
print("Credits: " + str(Credits))

# Correct indexing
Class = "INFO 648"
print(Class[0])

# Incorrect f-string formatting
print("I am enrolled in the {Program} program.")

# Correct f-string formatting
print(f"I am enrolled in the {Program} program.")
```

# PART 4: String Parsing:
String parsing is the process of analyzing a text string and extracting specific, meaningful information from it by identifying patterns, words, or characters.

Suppose we have this unstructured text:

"Order ID: 1025, Product: Laptop, Price: $800"

Using string parsing, we can extract individual pieces of information:
- Order ID → 1025
- Product → Laptop
- Price → $800

**Purpose**:To convert unstructured text into structured, usable information for further analysis.

A **regular expression (regex)** is a mini-language for describing a pattern. We use Python's built-in `re` module.

## Why This Matters in Data Analytics

**Where do we use strings?**

Strings appear in everyday business data, such as:
- Customer names and email addresses
- Phone numbers
- Order IDs (e.g., `ORD-2026-00457`)
- Dates and prices stored as text
- Customer feedback and survey responses

**What can go wrong?**

- **Inconsistent names:** "John", "john", and "John " may be treated as different customers.
- **Incorrect calculations:** "$1,200" is stored as text, so we cannot directly calculate the total.
- **Duplicate records:** Extra spaces or different capitalization can make the same customer appear more than once.
- **Matching problems:** "Alice" and "Alice " may not match when combining two datasets.



### Regex Symbols
| Symbol | Meaning |
|---|---|
| `\d` | one digit |
| `\d+` | one or more digits |
| `\w` | letter, digit or underscore |
| `\s` | whitespace (space, tab, newline) |
| `.` | any character |
| `[a-z]` | any one lowercase letter |
| `{4}` | exactly 4 times |
| `( )` | capturing group: "remember this part" |
| `r"..."` | a *raw string*: always put `r` before regex patterns |

```
import re  # Import Python's regular expression toolbox

# Sample order information
text = "Order ORD-2026-00457 placed on 2026-03-14 by alice@example.com for $1,200.50"

print(text)
```

##Regex Functions

### `re.search()` – find the first match anywhere in the text

```
# Find a date in the text
m = re.search(r"\d{4}-\d{2}-\d{2}", text)

print(m)          # Shows the match information
```

### `re.match()` – Check for a Match at the Beginning of Text

```
print(re.match(r"Order", text))  # Finds 'Order' because it is at the beginning
print(re.match(r"ORD", text))    # Returns None because 'ORD' is not at the beginning
print(re.search(r"ORD", text))   # Finds 'ORD' because it searches anywhere
```

### `re.findall()` – find all matches in a text

```
# Find all numbers in the text
numbers = re.findall(r"\d+", text)
print(numbers)
```

### `re.split()` – split text on a different seperators

```
messy = "red, green;blue   yellow"

# Split using only commas
print(messy.split(","))

# Split using commas, semicolons, or spaces
print(re.split(r"[,;\s]+", messy))
```

### `re.sub()` – substitute (find and replace a text by pattern)

```
# Replace every digit with #
print(re.sub(r"\d", "#", "Call 555-1234"))
```

### Capturing groups – Extract Specific Parts of a Match

```
# Extract the year and order number from the Order ID
m = re.search(r"ORD-(\d{4})-(\d+)", text)

print("Full Order ID:", m.group(0))
print("Year:", m.group(1))
print("Order Number:", m.group(2))
```

```
date = re.search(r"(\d{4})-(\d{2})-(\d{2})", text)

print(date.group(0))  # Full date
print(date.group(1))  # Year
print(date.group(2))  # Month
print(date.group(3))  # Day
```

### Type conversion
Everything extracted by regex is **text** (a `str`), even if it looks like a number.

```
# Extract the price
price_text = price = "1,200.50"

print(price_text, type(price_text))

# Remove the dollar sign and comma, then convert to float
price = float(price_text.replace("$", "").replace(",", ""))

print(price, type(price))

# Perform calculation
print(price * 2)
```

## ⚠️ Common mistakes in parsing
**Mistake 1:** Forgetting that extracted numbers are still text.

```
a = re.findall(r"\d+", "3 apples and 4 pears")
print(a)
print(a[0] + a[1])      # text glued together, NOT 7!
```

**Why:** `findall` returns strings, and `+` on strings joins them. **Fix:** convert with `int()`.

```
nums = [int(x) for x in a]
print(sum(nums))
```

**Mistake 2:** Assuming a match always exists.

```
# Search for a number in text that has no numbers
result = re.search(r"\d+", "no numbers here")

print(result.group())
```

```
#fix
result = re.search(r"\d+", "no numbers here")

if result:
    print(result.group())
else:
    print("No number found")
```

**Mistake 3:** Using the Wrong Search Function and Converting Text with Commas

```
# Example 1: float() cannot directly convert text containing commas

price = "1,200.50"

# This causes an error
try:
    print(float(price))
except ValueError:
    print("Error: Cannot convert text with commas into a number")

# Fix: Remove the comma before converting
price = float(price.replace(",", ""))
print("Correct price:", price)
```

## PART 5: String Cleaning

**String cleaning** is the process of fixing text by removing unwanted spaces, correcting capitalization, and removing unnecessary characters.

It helps make text consistent so that the same information is not treated as different values.

### Simple Example: Before and After Cleaning

| Before Cleaning | After Cleaning |
|---|---|
| `"  john smith "` | `"John Smith"` |
| `"NEW YORK "` | `"New York"` |
| `" Alice@gmail.com "` | `"alice@gmail.com"` |

**What changed?**
- Removed extra spaces.
- Made capitalization consistent.
- Standardized text formats.


##String Methods

### `strip()`, `lstrip()`, `rstrip()` – remove extra spaces

It helps in removing unwanted spaces from customer names, email addresses, and other text fields before analyzing data.

```
s = "   Alice Smith  \n"

# Display the original text
print(repr(s))

# Remove spaces from both ends
print(repr(s.strip()))

# Remove spaces from the left side
print(repr(s.lstrip()))

# Remove spaces from the right side
print(repr(s.rstrip()))
```

### `replace()` – swap exact text with new text

We use .replace() to remove or change unwanted characters in text.

```
print("$1,200".replace("$" , "").replace("," , ""))
```

### `lower()`, `upper()`, `casefold()` -Standardize text capitalization

```
# Convert text to lowercase
print("New York".lower())

# Convert text to uppercase
print("New York".upper())

# Compare two words using lower()
print("STRASSE".lower() == "Straße".lower())

# Compare two words using casefold()
print("STRASSE".casefold() == "Straße".casefold())
```

### Handling missing and empty strings

```
products = [" Laptop", "", "   ", None, "Keyboard"]

def clean_product(product):
    if product is None:
        return None

    product = product.strip()

    if product == "":
        return None

    return product

cleaned_products = [clean_product(p) for p in products]

print(cleaned_products)
```

### Removing duplicates

```
cities = ["New York", "Boston", "Chicago", "Boston", "New York"]

# Remove duplicates using set()
unique_cities = list(set(cities))

print(unique_cities)
```

## ⚠️ Common mistakes in cleaning

**Mistake 1**: Forgetting to save the cleaned string.

String methods do not change the original string. They return a cleaned version.

```
name = "  alice  "

# Mistake: The cleaned result is not saved
name.strip()
print(repr(name))

# Fix: Save the cleaned result back into the variable
name = name.strip()
print(repr(name))
```

**Mistake 2:** Counting duplicates before cleaning

```
emails = ["A@x.com", "a@x.com ", "a@X.COM"]

# Count unique emails before cleaning
print("Before cleaning:", len(set(emails)))

# Clean the emails
cleaned_emails = []

for email in emails:
    cleaned_emails.append(email.strip().lower())

# Display cleaned emails
print("Cleaned emails:", cleaned_emails)

# Count unique emails after cleaning
print("After cleaning:", len(set(cleaned_emails)))
```

**Mistake 3:** Calling a string method on a missing value.

```
customer_name = None

# Mistake: Trying to remove spaces from a missing value
try:
    print(customer_name.strip())
except AttributeError:
    print("Error: Cannot use strip() on a missing value")

# Fix: Check whether the value exists before cleaning
if customer_name is not None:
    customer_name = customer_name.strip()
else:
    print("Customer name is missing")
```

# Data Analytics Example

messy employee contact data  as it arrives in real life. We format, parse, then clean.

```
import re

# Step 1: Store the raw employee information
employee = "  Name: JOHN SMITH | Department: sales | Phone: (804) 555-1234 | Salary: $65,000  "

print("Raw Employee Data:", employee)
print("Data Type:", type(employee))


# Step 2: Format salary using an f-string
salary_number = 65000

print(f"\nEmployee Salary: ${salary_number:,}")


# Step 3: Extract phone number and salary using regex
phone = re.search(r"\(\d{3}\) \d{3}-\d{4}", employee)
salary = re.search(r"\$[\d,]+", employee)

print("\nExtracted Phone:", phone.group())
print("Extracted Salary:", salary.group())


# Step 4: Clean the employee information
clean_name = "  JOHN SMITH  ".strip()
clean_department = "sales".strip()
clean_phone = re.sub(r"\D", "", phone.group())
clean_salary = salary.group().replace("$", "").replace(",", "")


# Step 5: Convert salary from text to a number
salary_amount = int(clean_salary)

print("\nSalary as a Number:", salary_amount)
print("Data Type:", type(salary_amount))

# Step 6: Display the final clean employee record
print("\n--- Final Employee Record ---")
print("Name:", clean_name)
print("Department:", clean_department)
print("Phone:", clean_phone)
print(f"Salary: ${salary_amount:,}")
```

1. **Format** text and numbers so they are displayed consistently.
2. **Parse** text to extract useful information using string methods and regular expressions.
3. **Convert** extracted text into the correct data types, such as `int` or `float`.
4. **Clean** text by removing extra spaces, fixing capitalization, and removing unwanted characters.
5. **Check** for missing values and handle cases where a pattern is not found.
6. **Use clean data** for accurate analysis, comparisons, and reporting.

**Key takeaway:** Formatting, parsing, and cleaning help turn raw text into consistent and useful data for analysis.
