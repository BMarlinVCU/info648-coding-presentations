## Part 1: Loops

### What is a loop?

A loop allows Python to repeat the same action multiple times.

Instead of writing the same code over and over, we can use a loop to go through a group of items and perform an action on each item.

### Why Use a Loop?

Loops are useful when we need to repeat the same action for multiple items.

For example, in data analytics we may need to:
- Look at several values in a dataset.
- Perform the same calculation on multiple values.
- Check each item to see if it meets a condition.

A loop allows us to repeat these actions without writing the same code multiple times.

### Basic Structure of a For Loop

A `for` loop goes through each item in a group one at a time.

The basic structure is:

`for item in group:`

- `for` tells Python that we are starting a loop.
- `item` represents one item at a time.
- `in` tells Python where to get the items from.
- `group` is the collection of items we want to go through.
- `:` tells Python that the loop is beginning.

The code underneath the `for` statement is indented to show Python what should be repeated.

### Example 1: Using a Loop with Labels

First, we created a list called `fruits` that contains three items: apple, banana, and orange.

The loop then goes through the list one item at a time.

- First, `fruit` represents `"apple"` and Python prints apple.
- Next, `fruit` represents `"banana"` and Python prints banana.
- Finally, `fruit` represents `"orange"` and Python prints orange.

The loop stops after it has gone through every item in the list.

```python
fruits = ["apple", "banana", "orange"]

for fruit in fruits:
    print(fruit)
```

### Example 2: Using a Loop with Numbers

Loops can also be used to perform the same calculation on multiple values.

In this example, we have a list of sales amounts. We will use a loop to calculate a 10% increase for each amount.

For each amount, Python:
  1. Multiplies the amount by 1.10 to calculate a 10% increase.
  2. Stores the new amount in new_sale.
  3. Prints the new amount.

```python
sales = [100, 200, 300]

for sale in sales:
    new_sale = sale * 1.10
    print(new_sale)
```

### Common Mistake: Forgetting to Indent

Python uses indentation to determine which code belongs inside a loop.

A common mistake is forgetting to indent the code that should be repeated.

```python
fruits = ["apple", "banana", "orange"]

for fruit in fruits:
print(fruit)
```

### Why Did We Get an Error?

Python expected the `print()` statement to be indented because it is the action we want to repeat inside the loop.

Without the indentation, Python cannot determine what action belongs inside the loop.

To fix the error, we indent the `print()` statement.

```python
fruits = ["apple", "banana", "orange"]

for fruit in fruits:
    print (fruit)
```

### Example 3: Creating a New List with a Loop

A regular loop can also be used to create a new list.

In this example, we will start with a list of numbers and use a loop to create a new list containing each number multiplied by 2.

```python
numbers = [1, 2, 3, 4, 5]
doubled_numbers = []

for number in numbers:
    doubled_numbers.append(number * 2)

print (doubled_numbers)
```

### From Loops to List Comprehensions

We used a regular loop to create a new list of doubled numbers.

A list comprehension can also be used to create a new list, but it can often do the same task with less code.

Next, Caleb will use the same example to show us how a list comprehension compares to a regular loop.

## Part 2: List Comprehensions

### What is a list comprehension?

List comprehension is a concise way to create new lists by applying an expression to each item in an existing iterable like a list, tuple or range.

It helps to write clean, readable and efficient code compared to traditional loops.

### Why Use a List Comprehension?

While loops are useful when we need to repeat the same action for multiple items in a list, a list comprehension replaces the need to explicitly create an empty list, loop, and append.

For example, in data analytics we may need to:
- perform a calculation on numbers in a list ("square each number from an existing list")
- filter out certain items in a list ("only show even numbers from an existing list")
- identify characteristics of items in a list ("determine if each number from an existing list is odd or even")

A list comprehension allows us to perform these actions for the items in a list with one line of code.

### Basic Structure of a List Comprehension

A list comprehension creates a new list by evaluating an **expression** for each **item** in an **iterable**, and (if desired) keeps only items that satisfy a condition.

The basic structure is:

`[expression for item in iterable]`

- `expression` is an operation, label, or value to include in the new list.
- `for` tells Python that we are starting a loop.
- `item` is the current element from the iterable
- `in` tells Python where to get the items from.
- `iterable` is a sequence like a list, tuple, or range.

And if we want to add condition(s):

`[expression for item in iterable if condition]`
- `if condition` allows us to filter to only include items that satisfy the given condition.

##Example 4:  Using List Comprehension with Labels

So if we wanted to build off of the fruit example above, and transform the provided list (which is all lowercase) into all uppercase letters, we could use a list comprehension like the below:

```python
fruits = ["apple", "banana", "orange"]

uppercase = [fruit.upper() for fruit in fruits]
print (uppercase)
```

`fruit.upper()` = operation to perform (capitalize all letters)

`fruit` = item on which to perform operation

`fruits` = iterable (in this instance, a list)

##Common Error:

If your goal is just to print a list, list comprehension is not the right tool.  List comprehensions **create** lists, they don't then print that list as well.  Therefore, using a comprehension list to print a list doesn't make much sense.  

For example:

```python
[print  (fruit) for fruit in fruits]
```

We can see that it did indeed print each fruit from the list in above section, but it shows the list [None, None, None] as well.

If your goal is **printing**, a normal loop is a cleaner solution.  List comprehensions are meant to **build lists**, not print them.

### Example 5: Using a List Comprehension with Numbers

List comprehensions can also be used to perform the same calculation on multiple values.

In this example, we have a list of sales amounts. We will use a list comprehension to calculate a 10% increase for each amount.

```python
sales = [100, 200, 300]

new_sale = [sale * 1.10 for sale in sales]
print (new_sale)
```

Performing this action with a list comprehension saves space and is easier to read when compared to loop in Example 2 above.

**Loop version = Uses multiple lines to repeat the task and build a new list.**

**List comprehension version = Can perform the same task and build the new list in one line.**

##Example 6:

Finally, we can compare the structure of the loop above in Example 3 to a list comprehension.  Loop requires more lines and correct indentation, but list comprehension achieves same result with fewer lines and no indentation.

Note, though, that for multiple conditions / complex logic / long nested operations, it would probably be easier to understand the loop version v. the list comprehension version.

Loop:

```python
numbers = [1, 2, 3, 4, 5]
doubled_numbers = []

for number in numbers:
    doubled_numbers.append(number * 2)

print (doubled_numbers)
```

List Comprehension:

```python
numbers = [1, 2, 3, 4, 5]
doubled_numbers = [number * 2 for number in numbers]
print (doubled_numbers)
```
