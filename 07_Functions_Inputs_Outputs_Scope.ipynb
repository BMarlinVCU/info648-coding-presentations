# 2. User-defined Functions

User created functions are composed of the "def" keyword followed by a function name, parameters or arguments in parentheses, and a colon. You then write the statements of the function on following lines called the Function Body.

Many function bodies end with the last line using the "return" keyword or "print" function. Print just displays the output text to the user and is purely visual. Return actually hands over an output value to your program that you can then save as a variable or pass into other functions.

```python
def add_numbers(num1, num2):  #add_numbers is the function name and the paramaters/arguments are (num1, num2)
    result = num1 + num2  #this line is the function body, the code to be executed when the function is called
    return result #this line returns or prints the results of the function


sum_output = add_numbers(5, 7) #here we are saving the returned values of this function to the variable sum_output
print(f"The sum is: {sum_output}")
```

There are four main types of function arguments: Default, Keyword, Positional, and Arbitrary

Default Arguments use a predefined value when no value is passed during the function call.

```python
def greet(name, greeting="Hello"):
    print(f"{greeting}, {name}!")

# Calling the function without providing the optional argument
greet("Cara")

print() #this is to make the outputs look pretty

# Calling the function with a custom argument (overrides the default)
greet("Ryan", "Welcome")
```

Keyword Arguments pass values using parameter names, so argument order does not matter

```python
def student(fname, lname):
    print(fname, lname)

student(fname='John', lname='Tukey')

print()

student(lname='Tukey', fname='John') #since we call the parameter keywords, we can change the order of the arguments
```

Positional Argument values are assigned to parameters based on their order in the function call.

```python
def nameAge(name, age):
    print("Hi, I am", name)
    print("My age is ", age)


print("Correct positioning:")
nameAge("Olivia", 27)

print()

print("Incorrect Positioning:") #no error gets returned because 27 is mapped to name and "Olivia" is mapped to age
nameAge(27, "Olivia")
```

Arbitrary Arguments allow functions to accept multiple values, this is done using *args for positional arguments and **kwargs for keyword arguments. Arbitrary arguments are helpful when you do not know how many arguments will be passed into your function.

*args processes the positional arguments as a tuple

**kwargs processes the keyword arguments as a dictionary

```python
def list_students(*students):
    print("Here is your class roster:")
    for student in students:
        print(f"- {student}")

list_students("Ryan", "Cara", "Alex", "Josh", "Tim")
```

```python
def print_student_info(*courses, **personal_info):

    # *courses collects extra positional arguments into a tuple
    print("Courses Enrolled:")
    for course in courses:
        print(f"- {course}")

    print() # Blank line for spacing

    # **personal_info collects extra keyword arguments into a dictionary
    print("Student Details:")
    for key, value in personal_info.items():
        print(f"{key.capitalize()}: {value}")

# Calling the function with both types of arbitrary arguments
print_student_info("Math", "Physics", "Computer Science", name="Emily", grade="A", age=20)
```

# 3. Lambda Functions

###What is a Lambda Function?

A small anonymous function that can have any number of arguments, but can only execute a single function.

###Syntax
   `lambda arguments: expression`

   Does not require `def` or `return` keyword and does not need a name.


### Examples


Simple Squaring Lambda Function

```python
square = lambda x: x ** 2
print(square(5))
```

Multiple Arguments

```python
add = lambda a, b: a + b
print(add(3, 4))
```

Conditional Logic

```python
is_even = lambda n: "even" if n % 2 == 0 else "odd"
print(is_even(7))
```

Common Use : inline with map(), filter(), and sorted()

```python
nums = [1, 2, 3, 4, 5, 6]

doubled = list(map(lambda x: x * 2, nums))
evens   = list(filter(lambda x: x % 2 == 0, nums))

people = [("Marlin", 30), ("Bass", 25), ("Dolphin", 35)]
by_age = sorted(people, key=lambda p: p[1])
print(by_age)
```

# 4. Common Errors

Forgetting to `return`



```python
def add(a, b):
    result = a + b      # computed, but never returned

print(add(2, 3))
```

```python
#Fix : add the return statement
def add(a, b):
    return a + b

print(add(2, 3))
```

Incorrect Number of Statements

```python
def greet(name):
    return f"Hello, {name}"

greet()                   # TypeError: missing 1 required positional argument
greet("Buddy", "Pal")     # TypeError: takes 1 positional argument but 2 were given
```

```python
# Fix: pass exactly what the function expects, or add a default
def greet(name="friend"):
    return f"Hello, {name}"

print(greet())
print(greet("Pal"))
```

Forgetting the Parentheses when Calling

```python
def say_hi():
    return "hi"

print(say_hi)
print(say_hi())
```

Changing a Variable from Outside the Function (Scope)

```python
count = 0

def increment():
    count += 1
```

```python
# Fix : pass it in and return it
def increment(count):
    return count + 1

count = increment(count)
```

Mutable Default Arguments

```python
def add_item(item, items=[]):
    items.append(item)
    return items

print(add_item("a"))    # ['a']
print(add_item("b"))    # ['a', 'b']  <- surprise! The list is shared between calls

# Fix: use None as the default
def add_item(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items
```
